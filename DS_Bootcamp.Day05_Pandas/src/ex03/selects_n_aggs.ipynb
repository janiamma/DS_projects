```python
import pandas as pd
```

#### 1. Загрузить json файл

```python
df = pd.read_json('../data/auto.json', orient='records', encoding='utf-8')
```

```python
df.set_index('CarNumber', inplace=True)

df
```

#### 2. Сделать выборки

> - штрафы больше 2100

```python
df_fines = df[df['Fines'] > 2100]

df_fines
```

> - возврат равен 2, штрафы больше 2100

```python
df_refund = df_fines[df_fines['Refund'] == 2]
df_refund
```

> - строчки, содержащие в столбце моделей *Focus* и *Corolla*

```python
filter_models = ['Focus', 'Corolla']
df_models = df[df['Model'].isin(filter_models)]
df_models
```

> - выборка по номерам машин

```python
selected_numbers = ['Y7689C197RUS', '92928M178RUS', '7788KT197RUS', 'H115YO163RUS', 'X758HY197RUS']
df_numbers = df.loc[selected_numbers]
df_numbers
```

#### 3. Агрегации по *Make* и *Model*

- Медианный штраф

```python
df_make_median = df.groupby('Make')['Fines'].median()
print(df_make_median)
```

- Медианный штраф с группировкой по двум столбцам

```python
df_make_median2 = df.groupby(['Make', 'Model'])['Fines'].median()
print(df_make_median2)
```

- количество штрафов по группам

```python
df_fines_count = df.groupby(['Make', 'Model'])['Fines'].count()
```

```python
print(df_fines_count)
```

- мин и макс штрафы по группам

```python
df_min_fines = df.groupby(['Make', 'Model'])['Fines'].min()
print(df_min_fines)
```

```python
df_max_fines = df.groupby(['Make', 'Model'])['Fines'].max()
print(df_max_fines)
```

- стандартное отклонение по группам

```python
df_fines_std = df.groupby(['Make', 'Model'])['Fines'].std()
print(df_fines_std)
```

#### 4. Агрегации по номерам автомобилей

- номера автомобилей по количеству штрафов в порядке убывания

```python
df_cars_fines_desc = df.groupby(df.index)['Fines'].count().sort_values(ascending=False)
df_cars_fines_desc
```

- все строчки с номером авто **топ-1**

```python
df_top_car = df.loc[df_cars_fines_desc.index[0]]
df_top_car
```

- номера авто оплативших суммы штрафов в убывающем порядке

```python
df_cars_fines_sum_paid_desc = df.groupby(df.index)['Fines'].sum().sort_values(ascending=False)
df_cars_fines_sum_paid_desc
```

```python
df_top_car_paid = df.loc[df_cars_fines_sum_paid_desc.index[0]]
df_top_car_paid
```

```python
car_models = df.groupby(df.index)['Model'].nunique()
ambiguous_cars = car_models[car_models > 1]
ambiguous_cars
```

```python
# df.groupby(df.index)['Model'].unique()
df_more_models = df.loc[ambiguous_cars.index[0]]
df_more_models
```
